## 2. What kind of data is this?

Econometrics distinguishes four data structures, and which one you have decides what questions you can answer with it (this is week 1 material—the Essential Concepts videos have the definitions).

- Cross-sectional: many units, each observed once, at roughly one point in time.
- Pooled cross-section: cross-sections from several periods stacked together, but each unit still appears only once.
- Panel: the same units observed repeatedly over time.
- Time series: one unit observed at many points in time.

The decision tree, phrased as questions you ask the data:

1. Does each row represent a different unit at a single point in time? Likely cross-sectional.
2. Does the same unit appear in multiple rows over time? Panel or time series.
3. Multiple time periods, but no unit appears more than once? Likely pooled cross-section.
4. One single unit (one country, one firm) measured over time? Likely time series.

In a spreadsheet you'd answer these by scrolling and sorting.
In pandas, a few one-liners answer them faster, and they work the same whether the file has 200 rows or 20 million:

| Command | What it tells you |
|---|---|
| `df = pd.read_csv(url)` | load a dataset (straight from a link) into a dataframe called `df` |
| `df.head()` | the first five rows |
| `df.shape` | (number of rows, number of columns) |
| `df['col'].nunique()` | how many distinct values a column has |
| `df['col'].value_counts()` | each distinct value and how often it appears |
| `df['col'].min(), df['col'].max()` | smallest and largest values |

## 4. ⭐ Optional: from webpage to data in one line

A spreadsheet cannot do this one.
pandas can read every table on a webpage straight into dataframes:

```python
# ⭐ Read and run (needs internet)
tables = pd.read_html("https://en.wikipedia.org/wiki/List_of_cities_in_Australia_by_population",
                      match="Population", storage_options={"User-Agent": "Mozilla/5.0"})
cities = tables[0]
cities.head()
```

```python
# ⭐ Read and run: a one-line picture (plus one line to find the population column,
# because Wikipedia's column names change over time)
pop = [c for c in cities.columns if str(c).startswith("Population")][0]
cities.head(8).plot(x=cities.columns[1], y=pop, kind="barh", legend=False)
```
