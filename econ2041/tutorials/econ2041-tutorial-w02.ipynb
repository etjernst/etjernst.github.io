## 4. What kind of data?

As we saw in Week 1's Essential Concepts lectures, Econometrics usually considers four main data structures, and the type of data you have determines the types of questions that you can answer with it.

- **Cross-sectional:** many units, each observed once, at roughly one point in time
- **Pooled cross-section:** several cross-sections from different time periods "stacked" together, but each unit still appears only once
- **Panel:** the same units observed repeatedly over time
- **Time series:** one unit observed at many points in time

Questions to ask to figure out what type you are dealing with:
1. Does each row represent a different unit at a single point in time? $\rightarrow$ Likely cross-sectional
2. Does the same unit appear in multiple rows over time? $\rightarrow$ Panel or time series
3. Multiple time periods, but no unit appears more than once? $\rightarrow$ Likely pooled cross-section
4. One single unit (one country, one firm) measured over time? $\rightarrow$ Likely time series

In a spreadsheet, you'd answer these by sorting and/or scrolling.
In pandas, a few one-liner code snippets answer them faster (and they work whether the file has 200 rows or 20 million):

| Command | Description |
|---|---|
| `df = pd.read_csv(url)` | Load a dataset (from a URL, in this case) into a dataframe called `df` |
| `df.head()` | Show me the first five rows |
| `df.shape` | Show me (number of rows, number of columns) |
| `df['col'].nunique()` | Display how many distinct values a column has |
| `df['col'].value_counts()` | Count each distinct value and show me how often it appears |
| `df['col'].min(), df['col'].max()` | Show the smallest and largest values in the column, respectively |

### ⭐ From webpage to data in one line

In case you're still wondering what the big deal with Python is all about, here's something that a spreadsheet can't do: turn wikipedia pages into data!

One cool thing about pandas is that it can read tables on a webpage straight into a dataframe.

🤔 First, open this page in your normal browser: https://en.wikipedia.org/wiki/List_of_cities_in_Oceania_by_population

Look at the table, which contains a list of cities in Oceania with a population over 80,000. Imagine that you wanted to use that table as a dataset for analysis.

What would you actually do? Copy-paste it into Excel and clean it up by hand? Retype it?

Once you have a plan in mind, run the next cell.

```python
# ⭐ Read and run
tables = pd.read_html("https://en.wikipedia.org/wiki/List_of_cities_in_Oceania_by_population",  # grabs every table on the page
                      match="Urban area", storage_options={"User-Agent": "Mozilla/5.0"})        # match picks the cities table
cities = tables[0][["City", "Urban area"]]    # Keep city and population (for some reason, the population column is labeled "Urban area")
cities.columns = ["City", "Population"]       # Give them simple names
cities.head(10)

# The storage_options part is us telling pandas to introduce itself as a browser
# Wikipedia rejects anonymous scripts... so we make pandas pretend it's a browser!
```

```python
# ⭐ Read and run: a picture of the ten biggest 
cities.head(10).plot(x="City", y="Population", kind="barh", legend=False, ylabel="").invert_yaxis()
```
