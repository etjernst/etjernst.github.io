The notation comes from the Math review 1 Essential Concepts video: $Y_i$ is the income of household $i$, and $i$ runs from 1 to $n$, the number of households in the sample.

### The notation on three households

Suppose the sample held only three households, with incomes 100, 200, and 300.

- The total: $\sum_{i=1}^{3} Y_i = 100 + 200 + 300 = 600$
- The average: $\bar{Y} = \frac{1}{3}\sum_{i=1}^{3} Y_i = \frac{600}{3} = 200$

The $\sum$ sign says "add these up", the $i=1$ underneath says where to start counting, and the number on top says where to stop.
In the real dataset the number on top is $n$, so finding $n$ is where we start.

### Question 6: deviations from the mean

A deviation is how far one household sits from the average, $Y_i - \bar{Y}$.
Some households sit above it, some below.
The Essential Concepts video claims that these deviations always add up to exactly zero, whatever the data look like.

🤔 Suppose the richest household in this sample doubled its income overnight.
Would the deviations still add to zero, or would that one enormous household drag the sum above it?

```python
# ✏️ You write: add up every household's distance from the mean
# hint: (village['income'] - village['income'].mean()).sum()
```

Positive and negative deviations cancel, so their sum says nothing at all about how spread out incomes are.
Squaring each one first fixes that, since a squared number is never negative.

```python
# 🟢 Read and run: the sum of squared deviations
((village['income'] - village['income'].mean()) ** 2).sum()
```

The sum of squared deviations is the raw material for the variance and the standard deviation, which are next week's business.

## 3. Functional forms: what a log looks like

From here on, work at your own pace.
Your tutor will run this section with the class if there is time; otherwise it is yours to do afterwards, and Friday's answer key covers all of it.

A firm sells its product for 200 dollars a unit.
It buys advertising in units costing 1,000 dollars each, and its monthly budget stretches to at most 50 of them.
Advertising spend $A$ drives monthly sales $Q$ through

$$Q(A) = 500 \cdot \ln(A + 1)$$

Revenue is $R(A) = 200 \cdot Q(A)$, advertising costs $C(A) = 1000 \cdot A$, and profit is $\pi(A) = R(A) - C(A)$.

The $\ln$ is the natural log from the Math review 2 Essential Concepts video.
The $+1$ inside it is there because $\ln(0)$ is undefined, so it keeps the function alive at zero advertising.

| Command | Description |
|---|---|
| `np.arange(start, stop, step)` | Build evenly spaced numbers, stopping just before `stop` |
| `np.log(x)` | The natural log of `x` |
| `pd.DataFrame({'name': values})` | Build a dataframe from scratch |
| `df['new'] = ...` | Add a new column to a dataframe |
| `df['col'].idxmax()` | The row label where `col` reaches its highest value |

```python
# 🟢 Read and run: sales at six levels of advertising
ads = pd.DataFrame({'A': np.arange(0, 51, 10)})   # advertising spend: 0, 10, 20, 30, 40, 50
ads['Q'] = 500 * np.log(ads['A'] + 1)             # sales at each level of spend
ads
```

```python
# ✏️ You write: add a revenue column, 200 times sales
# hint: ads['R'] = 200 * ads['Q']
#       ads
```

```python
# ✏️ You write: add a profit column, revenue minus the cost of the advertising
# hint: ads['profit'] = ads['R'] - 1000 * ads['A']
#       ads
```

🤔 Read down the `Q` column.
Going from 0 to 10 buys some extra sales, and going from 40 to 50 buys some more.
Are those two increases anywhere near the same size?

Six points are too coarse to see the shape of anything.
Let us compute the same quantities on a fine grid, and push it well past the current budget so we can see where the function is heading.

```python
# 🟢 Read and run: the same functions, every half unit, out to A = 150
fine = pd.DataFrame({'A': np.arange(0, 150.5, 0.5)})
fine['Q'] = 500 * np.log(fine['A'] + 1)
fine['profit'] = 200 * fine['Q'] - 1000 * fine['A']
fine.head()
```

```python
# 🟢 Read and run: sales against advertising
plt.plot(fine['A'], fine['Q'])
plt.xlabel('Advertising spend, A')
plt.ylabel('Sales, Q')
plt.show()
```

```python
# 🟢 Read and run: profit against advertising
plt.plot(fine['A'], fine['profit'])
plt.axvline(50, linestyle='--', color='grey')   # the edge of the firm's current budget
plt.xlabel('Advertising spend, A')
plt.ylabel('Profit')
plt.show()
```

🤔 Look at the profit curve at the dashed line, where the budget runs out.
Is profit still climbing there?
Where does it stop climbing?

```python
# ✏️ You write: which level of advertising gives the highest profit?
# hint: fine.loc[fine['profit'].idxmax()]
```

## 4. ⭐ On your own: holding other things constant

A standard description of what a factory makes from capital $K$ and labor $L$:

$$Y = K^{0.3} L^{0.7}$$

Ceteris paribus means changing one input and leaving the other one alone.
That sounds like a Latin phrase you have to memorize; in code it is just a number you do not touch.

```python
# ⭐ Read and run: output from capital K and labor L
def output(K, L):
    return (K ** 0.3) * (L ** 0.7)

output(100, 100)
```

```python
# ✏️ You write: output when capital rises by 1 percent and labor stays where it is
# hint: output(101, 100)
```

🤔 Capital went up by 1 percent.
By what percent did output go up?
Now compare that to the exponent sitting on $K$.

### ⭐ One more log fact

The Essential Concepts video says $\ln(1 + x) \approx x$ when $x$ is small, which is why a log difference of 0.02 gets read straight off as "2 percent".

```python
# ⭐ Read and run: how good is that approximation?
np.log(1.02)
```

Close enough to 0.02 to be worth using.
It gets worse the bigger the change: compare `np.log(1.5)` against 0.5 and see how far apart they have drifted.
