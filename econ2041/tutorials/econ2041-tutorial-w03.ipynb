The notation comes from the Math review 1 Essential Concepts video: $Y_i$ is the income of household $i$, and $i$ runs from 1 to $n$, the number of households in the sample.

### Our notation: a three-household example

Suppose the sample held only three households, with incomes 100, 200, and 300.

- The total: $\sum_{i=1}^{3} Y_i = 100 + 200 + 300 = 600$
- The average: $\bar{Y} = \frac{1}{3}\sum_{i=1}^{3} Y_i = \frac{600}{3} = 200$

The $\sum$ sign says "add these up", the $i=1$ underneath says where to start counting, and the number on top says where to stop counting.
In the real dataset the number on top is $n$, so we will start by finding $n$.

Here are the commands that you'll need for this question.

| `df['col'].sum()` | Add up every value in the column |

### Question 6: deviations from the mean

A deviation is how far one household sits from the average, $Y_i - \bar{Y}$.
Some households sit above it, some below.
The Essential Concepts video claims that these deviations always add up to exactly zero, whatever the data look like.

🤔 Suppose the richest household in this sample doubled its income overnight.
Would the deviations still add to zero, or would that one enormous household drag the sum above it?

```python
# ✏️ You write: add up every household's distance from the mean
# hint: (village['income'] - mean_income).sum()
```

Positive and negative deviations cancel, so their sum says nothing at all about how spread out incomes are.
Squaring each one first fixes that, since a squared number is never negative.

```python
# 🟢 Read and run: the sum of squared deviations
((village['income'] - mean_income) ** 2).sum()
```

The sum of squared deviations is the raw material for the variance and the standard deviation, which are next week's business.

A survey in a rural NSW town interviews 1,250 working-age adults and records what each one was doing: employed, unemployed, or not in the labor force.
The following year the survey runs again, same size and same method, this time drawing its fresh sample of 1,250 from adults who were not interviewed the first time.

`nsw-labor-status.csv` contains both rounds, one row per person.

| Variable | Description |
|---|---|
| `id` | Person identifier |
| `year` | Survey round |
| `status` | `Employed`, `Unemployed`, or `Not in labor force` |

Three definitions:
`
- The **labor force** is everyone employed plus everyone unemployed; anyone who is neither sits outside it
- The **labor force participation rate** is the labor force divided by the working-age population
- The **unemployment rate** is the number unemployed divided by the labor force (note: NOT by the population)

| Command | Description |
