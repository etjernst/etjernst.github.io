### Question 4: splitting the math scores in two

So far, we've looked at `book` and `math` separately, let's examine whether and how they are related. 
For example, we might want to ask 

> "Are students with more books at home more likely to have higher math scores?"

To answer that, we can put the two variables in one table.

```python
# 🟢 Read and run: let's try to put book and math scores in a table 
# We use pd.crosstabs(var1, var2) to count the students in each "cell" 
# i.e. each combination var1 and var2, with var1 in the rows and var2 in the columns
# In our case, we want to see the combinations of values for pisa["book"] and pisa["math"]
pd.crosstab(pisa["book"], pisa["math"])
```

💬 Hmmm... ok, that's not very useful! What went wrong here? 

Let's think this through: 
- `book` has six categories
- `math` has 3,858 values (remember how to see this? `pisa['math'].nunique()` gives the number of unique values)... So cross-tabulating every possible value gives us an enormous table. 

So the table has exactly the information we asked for... but it's way too detailed for us to get an overview!

Let's simplify the question a bit: instead of keeping every possible math score, we can divide students into two groups:
1. at or below the median math score
2. above the median math score

```python
# 🟢 Read and run: split the students at the median math score
median_math = pisa["math"].median()

# np.where tells us: 
#   - where the test inside the parentheses is true, use the first label, 
#   - where the test inside the parentheses is false, use the second label
pisa["math_group"] = np.where(pisa["math"] > median_math, "Above median", "Below median")

print(f"median math score: {median_math}")
pisa[["student_id", "math", "math_group"]].head() # Look at the result
```

```python
# 🟢 Read and run: what share of students are above the median?
pisa["math_group"] == "Above median"
```

```python
# 🟢 Read and run: turn those True/False values into a share
(pisa["math_group"] == "Above median").mean()
```

### Question 5: the joint PMF

$P(X=x, Y=y)$ is the probability of each combination of values of two variables

Here, we want to know what fraction of students are in each combination of book range and math group?

Earlier, we used `pd.crosstab` to tally every combination of `book` and `math` and got a huge ugly table. Now that `math_group` only has two values, the same kind of table becomes more useful!

```python
# 🟢 Read and run: the joint counts
pd.crosstab(pisa["book"], pisa["math_group"])
```

### Question 6: getting the original PMF back

- The joint table above contains probabilities for `book` and `math_group` together
- Suppose we stop caring about math group, can we recover the probabilities for `book` from this table?

So far, our probabilities have used all 12,136 students as the reference group.
Now suppose we learn something about a student before asking our probability question, and we want to calculate the probability that they scored above the median in math.

Start with the first row of the counts table from question 5:

```python
# ✏️ You write: among students with 0-10 books, what share scored above the median?
# Hint: 56 of the 446 students scored above the median
```

We can now ask two more questions about the same 56 students:

- **Joint probability:** out of all 12,136 students, what share have 0-10 books **and** score above the median?
- **Conditional probability:** out of the 446 students with 0-10 books, what share score above the median?

The numerator is 56 in both cases, we only change the denominator

```python
# 🟢 Read and run: same 56 students, two different reference groups
print("Joint probability:", round(56 / 12136, 3))
print("Conditional probability:", round(56 / 446, 3))
```

Staying with the same group of 446 students, we also know that 390 of them scored below the median...

```python
# ✏️ You write: among students with 0-10 books, what share scored below the median?
```

Those two numbers together are the **conditional distribution of math group, given 0-10 books**.

We could repeat the same calculation for every book range by asking:

> "Within this book range, what share of students are above the median and what share are below?"

Rather than doing all twelve divisions by hand, we can have `pd.crosstab` do them for us!

```python
# 🟢 Read and run: within each book range, find the share in each math group
pd.crosstab(pisa["book"], pisa["math_group"], normalize="index").round(3)
```

## 3. Are books at home independent of math scores?

Recall that two variables are independent if conditioning on one tells us nothing about the other:

$$P(A \mid B) = P(A) \quad \text{for every value of } B$$

We already know that about half of all students score above the median.

If books at home and math scores were independent, knowing a student's book range should not change that probability. The share above the median should still be about 0.50 in every row of our conditional-distribution table.

So let's look down the "Above median" column and see whether it stays at 0.50.

### Question 9: what does independence actually look like?

In our data, the probability of scoring above the median changes a lot across book ranges.

🤔 What would these conditional probabilities look like if books at home and math performance really were independent?

Let's build a world where they are independent: 
- keep every math score where it is
- reshuffle the `book` column and "distribute" it randomly

```python
# 🟢 Read and run: a world where books and math are independent by construction
shuffled = pisa.copy()
shuffled["book"] = pisa["book"].sample(frac=1, random_state=1).values

pd.crosstab(shuffled["book"], shuffled["math_group"], normalize="index").round(3)
```
