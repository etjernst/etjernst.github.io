## Variables

A variable stores a value so you can use it later.
The name goes on the left, the value on the right, and `=` assigns one to the other.

Variable names cannot contain spaces.

```python
# 🟢 Read and run
x = 5
name = "Emilia"

print(x)
print(name)
```

## Strings

A string is any piece of text.
It has to be wrapped in quotes, either `'single'` or `"double"`.

```python
# 🟢 Read and run
country = "Australia"
print(country)
```

## Numbers

Python has two kinds of numbers: integers, which are whole, and floats, which have a decimal point.

```python
# 🟢 Read and run
a = 4      # an integer
b = 5.2    # a float

print(a)
print(b)
```

## Python is a calculator

The symbols are what you would expect: `+` and `-` do the obvious, `*` multiplies, and `/` divides.

Variables work anywhere a number would.

```python
# 🟢 Read and run
print(3 + 5)

five = 5
y = 2

print(y + five)
print(y / five)
print((y / five) * 10)
```

## Booleans

A boolean is either `True` or `False`, and both start with a capital letter.

You get one back whenever you ask Python a yes-or-no question.

```python
# 🟢 Read and run
is_student = True
age = 17

print(is_student)
print(age >= 18)
```

## Checking types

When you are not sure what you are holding, `type()` will tell you.

```python
# 🟢 Read and run
print(type(name))
print(type(a))
print(type(b))
print(type(is_student))
```

## Adding strings together

`+` works on strings too, but it joins them rather than adding them up.
Notice the space at the end of the first one.

```python
# 🟢 Read and run
first = "Econometrics "
second = "is "
third = "awesome!"

print(first + second + third)
```

## Lists

A list stores several values in order, inside square brackets and separated by commas.

You get an item back by its position, and Python starts counting at zero.

```python
# 🟢 Read and run
countries = ["Kenya", "Nigeria", "Ghana", "Malaysia", "Indonesia"]

print(countries[0])
print(countries[1])
```

## Dictionaries

A dictionary stores pairs: a key, and the value that belongs to it.
Curly brackets define it, square brackets retrieve from it.

```python
# 🟢 Read and run
gdp = {"Kenya": 110, "Nigeria": 440}

print(gdp["Kenya"])
```
